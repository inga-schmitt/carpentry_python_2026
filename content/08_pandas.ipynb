```python
# Alle Daten in einem Dataframe zusammenführen (vgl. Übung 8)

# importiere Module
import glob
import pandas as pd

dfs = []   # initializiere neue (leere) Liste

for csv in sorted(glob.glob('data/*_circ.csv')):
    year_number = csv[5:9]                  # extrahiere Jahreszahl aus Dateinamen
    data = pd.read_csv(csv) 
    data['year_header'] = year_number       # generiere in jedem einzelnen Datensatz neue Spalte einfach durch Zuweisung
    dfs.append(data)                        # jeder einzelne Datensatz wird an die Liste angehängt

df = pd.concat(dfs, ignore_index=True)      # mit concat wird aus der Liste ein neuer einheitlicher Datensatz generiert

df.head(3)    # erste 3 Zeilen
```

```python
# Wie viele Zweigstellen gibt es?

df['branch'].unique()   # finde alle unterschiedlichen Werte (jeden nur einmal ausgeben, auch wenn mehrfach vorhanden)
```

```python
# Teildatensatz betrachten

df.groupby('branch')['ytd'].sum()      # gruppieren nach Kriterium: für jede Zweigstelle (branch) über alle Jahre 
                                       # (impliziert, weil ein großer Datensatz aus den Jahresdatensätzen) die Spalte 'ytd' aufsummiert
```

```python
# Sortieren

circ_by_branch = df.groupby('branch')['ytd'].sum()                    # wie oben, Variable zugewiesen
circ_by_branch_sorted = circ_by_branch.sort_values(ascending=False)   # sortieren in absteigender Reihenfolge
circ_by_branch_sorted.head(10)                                       # die ersten 10 Zeilen anzeigen
```

```python
# gruppieren nach mehreren Spalten

circ_by_year_branch = df.groupby(['year_header','branch'])['ytd'].sum().sort_values(ascending=False)    # Gruppierung nach Jahr und Zweigstelle in [] 
circ_by_year_branch.head(10)
```

```python
# Nachtrag: Auswahl mit 2-dimensionalem numerischen Index: iloc

# iloc nimmt Indizes [Zeile, Spalte]
# hier 1. Zeile von 1. und vorletzter Spalte

print(f"Branch: {df.iloc[0,0]} \nYTD circ: {df.iloc[0,-2]}")
```

```python
# Auswahl mit Zeilenindex und Spaltenname: loc

print(f"Branch: {df.loc[0,'branch']} \nYTD circ: {df.loc[0, 'ytd']}")
```

```python
type(circ_by_branch_sorted)    # gefilterter Subdatensatz ist aktuell eine Datenreihe (Series)
```

```python
# wieder zu Dataframe machen

circ_df = circ_by_branch_sorted.to_frame()
circ_df.head(5)
```

```python
# speichern als csv

circ_df.to_csv("circ_by_branch.csv")
```
