## Übung 2: Variablenzuweisung

```python
x = 1.0
y = 3.0
swap = x  #  x = 1.0  y = 3.0  swap = 1.0
x = y     #  x = 3.0  y = 3.0  swap = 1.0
y = swap  #  x = 3.0  y = 1.0  swap = 1.0
print(x, y)
```

```python
x = 1.0
y = 3.0 # x = 1.0  y = 3.0
x = y   # x = 3.0  y = 3.0
y = x   # x = 3.0  y = 3.0
print(x, y)
```

## Übung 3: Werte vorhersagen

```python
initial = "left"  # Initial is assigned the string "left"
position = initial  # Position is assigned the variable initial, currently "left"
initial = "right"  # Initial is assigned the string "right"
print(position)
```

## Übung 5: Fortgeschrittene Slices

```python
library_name = 'LibraryOfBabel'
f'library_name[1:3] is: {library_name[1:3]}'
```

```python
print(f'library_name[7:] is: {library_name[7:]}')  #(ohne Wert nach dem Doppelpunkt)
print(f'library_name[:7] is: {library_name[:7]}')  #(ohne Wert vor dem Doppelpunkt)
print(f'library_name[:] is: {library_name[:]}')  #(nur der Doppelpunkt)
print(f'library_name[7:-5] is: {library_name[7:-5]}')  #(mit einer negativen Zahl)
```

## Übung 8: Länge und Indexes von Listen

```python
colors = ['red', 'blue', 'green']
print(len(colors))
print(colors[0])
```

## Übung 9: Slices von Listen

```python
numbers = [1, 2, 3, 4, 5, 6]
print(numbers[0:3])
print(numbers[3:6])
```

```python
numbers = [1, 2, 3, 4, 5, 6]
print(numbers[:3])
print(numbers[3:])
```

## Übung 10: Lückentext

```python
values = []
values.append(1)
values.append(3)
values.append(5)
print(f'first time: {values}')
values = values[1:3]
print(f'second time: {values}')
```
