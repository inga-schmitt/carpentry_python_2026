```python
# Ausgabe von Werten in einer Liste - ohne Schleife

odds = [1, 3, 5, 7]  # Listenerstellung
print(odds)
```

```python
# jedes einzelne Element mit jeweiligem Index

print(odds[0], odds[1], odds[2], odds[3])
```

```python
# mit Schleife

odds = [1, 3, 5]
 
for num in odds:       # Doppelpunkt, sonst funktioniert es nicht; Variable 'num' ist nur der Container, könnte anders heißen
    print(num)
```

```python
print(num)   # Variable bleibt bestehen nach Ende des Loops
```

```python
# Schleifen über alles, was iterierbar ist, auch strings (jedes Zeichen einzeln, Leerzeichen inklusive)

for letter in 'Bibliothek zu Babel':
    print(letter)
```

```python
# Übung 3 - Gemüse

veggies = ['Spinat', 'Karotte', 'Zucchini']
for veg in veggies:
    print(veg)
```

```python
# Übung 5 - string index

# "erst das Problem lösen, dann coden"
# collection: verschiedene Farben
# neuer String, Akronym
# erster Buchstabe --> [0]
# zusammenbauen mit +
# in jeder Umdrehung soll ein Buchstabe dazu kommen

acronym = ''  # neuer leerer String

col = ['red', 'green', 'blue']

for color in col:
    acronym = acronym + color[0]  # Zuweisung überschreibt den Wert der Variablen mit neuem Wert
    # acro = '' + 'r'             # 1. Runde
    # acro = 'r' + 'g'            # 2. Runde
    # acro = 'rg' + 'b'           # 3. Runde

print(acronym)
```
