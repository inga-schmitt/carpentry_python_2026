```python
# Schleife über eine Liste mit Dateinamen

import pandas as pd

for filename in ['data/2011_circ.csv', 'data/2012_circ.csv']:     # Liste händisch mit einzelnen Dateipfaden
    df = pd.read_csv(filename)
    print(filename, df['ytd'].max())    # Spalte 'ytd' (year to date, Summe übers ganze Jahr) des dataframes, davon der Maximalwert
```

```python
# globbing = pattern matching
# '*' null oder mehr Zeichen, Platzhalter (wildcard)
# '?' genau ein Zeichen

import glob  # glob Modul
# from glob import glob   # Alternative: direkt die Funktion importieren

glob.glob('data/*_circ.csv')  # erstellt Liste mit allen Pfaden, die zu dem Muster passen; glob.glob ist die Funktion des Moduls glob
```

```python
help(glob)
```

```python
# Schleife + globbing

filelist = sorted(glob.glob('data/*_circ.csv'))   # + Sortierfunktion

for filename in filelist:
    data = pd.read_csv(filename)
    print(filename, data['ytd'].max())
```

```python
# Übung 7: Minimale Zirkulation pro Jahr

import pandas as pd
for csv in sorted(glob.glob('data/*_circ.csv')):
    data = pd.read_csv(csv)
    print(csv, data['ytd'].min())
```

```python
# Übung 8: Dataframes zusammenführen

import pandas as pd

dfs = []

for csv in sorted(glob.glob('data/*_circ.csv')):
    data = pd.read_csv(csv)
    dfs.append(data)
    
new_df = pd.concat(dfs, ignore_index=True)
```
